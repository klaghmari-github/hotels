## Notes - Application Métier ROD-IA
L'app comprend le métier du coin de vente dans les hôtels Accor :

- **Inputs** : via les Écrans du simulateur ROD (variables descriptives : nb chambres, équipements, profil clients, corner...).
- **ML** : prédit le profil détaillé des ventes (286 cibles : par mois, F&B/NON, par GAMME) à partir des caractéristiques de l'hôtel.
- **Cohérence & What-if** : 
  - Le total CA est la base.
  - Répartition par GAMME visible.
  - Si le directeur dit "plus d'alcool" ou "je veux 10% F&B et 0% alcool", on **réalloue proportionnellement** aux autres catégories selon leur attractivité originale (parts dans la prédiction = proxy conversion).
  - Cela simule "libérer de l'espace" et "à qui ça profite ?".
- **Simulation Métier** : 
  - Funnel clients -> acheteurs (formules ROD).
  - Revenue ajusté par m_lin + mix.
  - Coûts (agencement, techno, annexes).
  - Marge, P&L.
  - Recommandations : meilleur concept + m_lin pour la marge.
- **Bouton "Full Business Sim"** : fait tout ça d'un coup et propose les 3 meilleures options.

Lance le serveur, va sur localhost:5000, remplis les écrans, prédis, joue avec les % et exclusions.
