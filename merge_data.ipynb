```python
rod_df = pd.read_excel("rod_prepared_data.xlsx", sheet_name=0)
rod_df = rod_df.drop(["HOTEL_LAT", "HOTEL_LON"], axis = 1)

print(rod_df.shape)
rod_df
```

```python
poi_df = pd.read_excel("poi_prepared_data.xlsx", sheet_name=0)
print(poi_df.shape)
poi_df
```

```python
weather_df = pd.read_excel("weather_prepared_data.xlsx", sheet_name=0)
print(weather_df.shape)
weather_df.head(3)
```

```python
transaction_df = pd.read_excel("transaction_prepared_data.xlsx", sheet_name=0)
print(transaction_df.shape)
transaction_df
```

```python
# Arrondi des coordonnées
poi_df["HOTEL_LAT"] = poi_df["HOTEL_LAT"].round(5)
poi_df["HOTEL_LON"] = poi_df["HOTEL_LON"].round(5)

weather_df["HOTEL_LAT"] = weather_df["HOTEL_LAT"].round(5)
weather_df["HOTEL_LON"] = weather_df["HOTEL_LON"].round(5)

transaction_df["HOTEL_LAT"] = transaction_df["HOTEL_LAT"].round(5)
transaction_df["HOTEL_LON"] = transaction_df["HOTEL_LON"].round(5)



# =========================
# Merge POI + WEATHER
# =========================
poi_weather_df = poi_df.merge(
    weather_df,
    on=["HOTEL_LAT", "HOTEL_LON"],
    how="inner"
)

poi_weather_df = poi_weather_df.drop(["HOTEL_LAT", "HOTEL_LON"], axis = 1)

# df_final = poi_df.merge(
#     weather_df,
#     on=["HOTEL_LAT", "HOTEL_LON"],
#     how="inner"
# )



# =========================
# Merge avec SALES
# =========================

# df_final = poi_weather_df.merge(
#     transaction_df,
#     on=["HOTEL_NAME", "HOTEL_LAT", "HOTEL_LON"],
#     how="inner"
# )

poi_weather_transaction_df = poi_weather_df.merge(
    transaction_df,
    on=["HOTEL_NAME"],
    how="inner"
)

# =========================
# Merge avec ROD
# =========================
# df_final = poi_weather_transaction_df.merge(
#     rod_df,
#     on=["HOTEL_NAME"],
#     how="inner"
# )

poi_weather_transaction_rod_df = poi_weather_transaction_df.merge(
    rod_df,
    on=["HOTEL_NAME"],
    how="inner"
)

merged_df = poi_weather_transaction_rod_df
merged_df = merged_df.fillna(0)
```
